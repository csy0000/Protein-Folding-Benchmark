# Inspect CASP15/CASP16 Target Manifest

QC views for `data/casp15_casp16_target_manifest.csv`.

```python
from pathlib import Path
import pandas as pd

manifest = Path('../data/casp15_casp16_target_manifest.csv')
df = pd.read_csv(manifest)
df.head()
```

```python
for column in ['casp_round', 'should_use', 'stoichiometry', 'type', 'chain_assignment_confidence']:
    display(df[column].fillna('').value_counts().rename_axis(column).reset_index(name='count'))
```

```python
yes = df[df['should_use'] == 'Yes']
display(yes)
```

```python
check = df[df['should_use'] == 'Check']
display(check[['casp_round', 'domain_id', 'pdb_id', 'chain_id', 'target_length', 'mapping_identity', 'mapping_coverage', 'reference_coverage', 'reason']])
```

```python
failures = df[df['chain_assignment_confidence'].isin(['failed', 'low'])]
display(failures[['casp_round', 'domain_id', 'pdb_id', 'target_length', 'chain_assignment_confidence', 'reason']])
```

```python
ax = df.assign(target_length=pd.to_numeric(df['target_length'], errors='coerce')).boxplot(column='target_length', by='should_use')
ax.set_title('Target length by QC state')
ax.figure.suptitle('')
```

```python
strict_columns = ['domain_id', 'target_id', 'pdb_id', 'chain_id', 'sequence', 'target_length', 'sequence_start', 'sequence_end', 'residue_start', 'residue_end']
strict = df[df['should_use'] == 'Yes'][strict_columns]
strict_out = Path('../data/casp15_casp16_strict_monomer_targets.csv')
strict.to_csv(strict_out, index=False)
strict_out, len(strict)
```

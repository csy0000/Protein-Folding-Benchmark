# Inspect prediction tables
Set RUN_ROOT to a benchmark result directory containing tables/per_model_metrics.csv.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RUN_ROOT = Path('../results/casp15_casp16_unique_lt1000_all_default_20260529_182536')
TABLES = RUN_ROOT / 'tables'
FIGURES = RUN_ROOT / 'figures'
FIGURES.mkdir(parents=True, exist_ok=True)
```

```python
metrics = pd.read_csv(TABLES / 'per_model_metrics.csv')
summary = pd.read_csv(TABLES / 'model_summary.csv')
target_summary = pd.read_csv(TABLES / 'target_summary.csv')
failed = pd.read_csv(TABLES / 'failed_predictions.csv')
metrics.head()
```

## Model summary by QC subset


```python
summary
```

## Score distributions by model and QC status


```python
for metric in ['lddt_ca', 'gdt_ts']:
    if metric in metrics and metrics[metric].notna().any():
        ax = metrics.boxplot(column=metric, by=['model', 'should_use'], rot=90, figsize=(12, 5))
        ax.figure.suptitle('')
        ax.set_title(metric)
        ax.set_ylabel(metric)
        ax.figure.tight_layout()
        ax.figure.savefig(FIGURES / f'{metric}_by_model_qc.png', dpi=150)
```
